```python
compteur = 1
if compteur == 1:
  # Get the base URL
  url = f"https://www.chucknorrisfacts.fr/facts/top/1"
  # Fetch data from the site
  site = requests.get(url).text
  # Parse the HTML
  soup = BeautifulSoup(site, "html.parser")
  print(f"\n Page {compteur} ------------------------------------------------- \n")
  for fact in soup.find_all("p", class_="card-text"):
    print(fact.text)
  # Increment the page counter on each iteration
  compteur = compteur + 1
else: # For subsequent pages
  # Find the page links
  for element in soup.find_all("a", class_="page-link"):
    # Target the "next" button containing the text >>
    if element.text == ">>":
      # Get the partial URL from this button (it only contains /facts/top/theNumber)
      # Reconstruct it to form the full URL
      url2 = "https://www.chucknorrisfacts.fr" + element.get("href")
      # Do the same process again
      site = requests.get(url).text
      soup = BeautifulSoup(site, "html.parser")
      print(f"\n Page {compteur} ------------------------------------------------- \n")
      for fact in soup.find_all("p", class_="card-text"):
        print(fact.text)
      compteur = compteur + 1
```
